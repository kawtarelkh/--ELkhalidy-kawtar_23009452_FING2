# 1. Cadrage de la question (00–10 min)

## 1.1 Contexte financier

Les anticipations d'inflation représentent les perceptions des agents économiques concernant l'évolution future des prix. Elles sont utiles pour comprendre les décisions de consommation, d'épargne, d'investissement et les anticipations de politique monétaire.

## 1.2 Problématique proposée

**Dans quelle mesure les observations mensuelles passées des anticipations d'inflation permettent-elles de prévoir leur valeur au mois suivant ?**

Cette formulation doit être confirmée ou ajustée selon la définition exacte de la série et les variables réellement disponibles.

## 1.3 Objectif mesurable

Construire un modèle de régression qui prédit la valeur de l'indicateur d'anticipations d'inflation au mois suivant, puis comparer ses performances à une référence naïve fondée sur la dernière observation connue.

- **Variable cible envisagée :** valeur de l'indicateur au mois \(t+1\).
- **Horizon :** un mois à l'avance.
- **Métrique principale :** MAE.
- **Métrique secondaire :** RMSE.

## 1.4 Hypothèses à examiner

1. Les valeurs récentes de l'indicateur contiennent potentiellement une information utile sur sa valeur future.

## 5.2 Référence naïve obligatoire

Pour une prévision à un mois, une référence naïve possible est la persistance :

\[
\widehat{Y}_{t+1}^{\,naïf}=MICH_t
\]

Elle prédit pour le mois suivant la dernière valeur connue. Vérifier que cette référence est alignée exactement sur les mêmes dates de test que le modèle.

## 5.3 Métriques

**MAE — erreur absolue moyenne :**

\[
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat y_i|
\]

**RMSE — racine de l'erreur quadratique moyenne :**

\[
RMSE=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}
\]

La MAE est directement interprétable dans l'unité de la cible. La RMSE pénalise davantage les erreurs importantes.
